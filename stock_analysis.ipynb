Cummulative Returns, Drawdowns etc..

```python
apple=yf.download("AAPL")
```

```python
apple=apple.Close.to_frame()
apple
```

```python
#daily return
apple["Daily_returns"]=np.log(apple.div(apple.shift(1)))
apple
```

```python
apple.dropna(inplace=True)
apple
```

```python
apple.Daily_returns.sum()
```

```python
np.exp(apple.Daily_returns.sum())#if we invest 1 dollar in apple stock, how much money we will have right now
```

```python
apple["Cummreturns"]=apple.Daily_returns.cumsum().apply(np.exp)
```

```python
apple
```

```python
apple.Cummreturns.plot(figsize=(12,8),fontsize=15)
plt.legend(loc="upper left",fontsize=15)
plt.title("Buy and Hold",fontsize=15)
plt.show()
```

```python
apple.Daily_returns.mean()*252
```

```python
apple.Daily_returns.std()*np.sqrt(252)
```

```python
#how to Calulate drawdowns
apple["Cummmax"]=apple.Cummreturns.cummax()
```

```python
apple
```

```python
apple[["Cummreturns","Cummmax"]].plot(figsize=(12,8),title="Apple Buy and Hold+ cummmax",fontsize=12)
plt.show()
```

```python
#apple["Drawdown"]=(apple.Cummreturns-apple.Cummmax)/apple.Cummmax
apple["Drawdown"]=apple["Cummmax"]-apple["Cummreturns"]
```

```python
apple
```

```python
apple.Drawdown.max()
```

```python
apple.Drawdown.idxmax()
```

```python
apple.loc[(apple.index<="2023-01-05")]#== pota antha date kulla data matum varum
```

```python
#finding percentage
apple["Drawdown%"]=(apple["Cummmax"]-apple["Cummreturns"])/apple["Cummmax"]
```

SMA strategy

```python
data=apple.Close.loc[(apple.index>="1991-01-01")]
```

```python
data=data.to_frame()
data
```

```python
sma_s=50
sma_l=100
```

```python
data["SMA_S"]=data.Close.rolling(sma_s).mean()
data["SMA_L"]=data.Close.rolling(sma_l).mean()
```

```python
data
```

```python
data.plot(figsize=(12,8),title="APPLE - SMA{} | SMA{}".format(sma_s,sma_l),fontsize=12)
```

```python
#to get specific year data
data.loc["2024"].plot(figsize=(12,8),title="APPLE - SMA{} | SMA{}".format(sma_s,sma_l),fontsize=12)
```

```python
data.dropna(inplace=True)
data
```

```python
data["Position"]=np.where(data.SMA_S>data.SMA_L,1,-1)
data
```

```python
#for all data
data.loc[:,["SMA_S","SMA_L","Position"]].plot(figsize=(12,8),title="APPLE - SMA{} | SMA{}".format(sma_s,sma_l),fontsize=12,secondary_y=["Position"])
```

```python

data.loc["2024",["SMA_S","SMA_L","Position"]].plot(figsize=(12,8),title="APPLE - SMA{} | SMA{}".format(sma_s,sma_l),fontsize=12,secondary_y=["Position"])
```

```python
data["ReturnsB&H"]=np.log(data.Close.div(data.Close.shift(1)))
data
```

```python
data["Strategy"]=data["ReturnsB&H"]*data.Position.shift(1)
```

```python
data
```

```python
data.dropna(inplace=True)
data
```

```python
data[["ReturnsB&H","Strategy"]].sum()
```

```python
data[["ReturnsB&H","Strategy"]].sum().apply(np.exp)#what 1$ will be
```

```python
data[["ReturnsB&H","Strategy"]].std()*np.sqrt(252)#annual std
```

```python
data[["ReturnsB&H","Strategy"]].cumsum().plot(figsize=(12,8),title="APPLE - SMA{} | SMA{}".format(sma_s,sma_l),fontsize=12)
```

## Strategy adjusted with long bias

```python
data["Position2"]=np.where(data.SMA_S>data.SMA_L,1,0)
data
```

```python
data["Strategy2"]=data["ReturnsB&H"]*data.Position2.shift(1)
```

```python
data.dropna(inplace=True)
```

```python
data[["ReturnsB&H","Strategy2"]].sum()
```

```python
data[["ReturnsB&H","Strategy2"]].sum().apply(np.exp)#what 1$ will be
```

```python
data[["ReturnsB&H","Strategy2"]].std()*np.sqrt(252)#annual std
```
